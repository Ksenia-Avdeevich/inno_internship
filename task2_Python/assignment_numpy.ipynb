```python
import numpy as np
```

**1.** Дан случайный массив, поменять знак у элементов, значения которых между 3 и 8

```python
# создаем случайный массив
n = 15 # кол-во элементов в массиве
a, b = 0, 20 #диапазон массива
arr = np.random.randint(a, b, size=n)

print(arr)
```

```python
cond = (arr > 3) & (arr < 8)
new_arr = np.where(cond, -arr, arr)

print(new_arr)
```

**2.** Заменить максимальный элемент случайного массива на 0

```python
new_arr = np.where((new_arr == new_arr.max()), 0, new_arr)
print(new_arr)
```

**3.** Построить прямое произведение массивов (все комбинации с каждым элементом). На вход подается двумерный массив

```python
# перестраиваем массив в двумерный
d_arr = new_arr[:4].reshape(2, 2)
print(d_arr, end="\n\n")

grid = np.meshgrid(*d_arr)
res_3 = np.stack(grid, axis=-1).reshape(-1, len(d_arr))
print(res_3)
```

**4.** Даны 2 массива A (8x3) и B (2x2). Найти строки в A, которые содержат элементы из каждой строки в B, независимо от порядка элементов в B

```python
arr_a = np.random.randint(a, b, size=24).reshape(8, 3)
print(arr_a, end="\n\n")
arr_b = np.random.randint(a, b, size=4).reshape(2, 2)
print(arr_b, end="\n\n")

mask_4 = np.array([np.isin(arr_a, row_b).any(axis=1) for row_b in arr_b]).all(axis=0)
res_4 = arr_a[mask_4]
print(res_4)
```

**5.** Дана 10x3 матрица, найти строки из неравных значений (например строка [2,2,3] остается, строка [3,3,3] удаляется)

```python
arr_10_3 = np.random.randint(a, 2, size=(10, 3))
print(arr_10_3, end="\n\n")

# mask = np.array([(row[0] != row[1] or row[1] != row[2]) for row in arr_10_3])

mask_5 = arr_10_3.min(axis=1) != arr_10_3.max(axis=1)
res_5 = arr_10_3[mask_5]
print(res_5)
```

**6.** Дан двумерный массив. Удалить те строки, которые повторяются

```python
print(arr_10_3, end="\n\n")

sort_un, ind = np.unique(arr_10_3, return_index=True, axis=0)
res_6 = sort_un[np.argsort(ind)]
print(res_6)
```

```python
arr = [[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]
ans = 1
for i in range(min(len(arr), len(arr[0]))):
    if arr[i][i] != 0:
        ans *= arr[i][i]

print(ans)
```

* __Задача 4__: Реализовать кодирование длин серий (Run-length encoding). Для некоторого вектора x необходимо вернуть кортеж из двух векторов одинаковой длины. Первый содержит числа, а второй - сколько раз их нужно повторить.  
 Например, для x = np.array([2, 2, 2, 3, 3, 3, 5]) ответ (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
x_z4 = [2, 2, 2, 3, 3, 3, 5]
ans1_z4 = set(x_z4)
ans2_z4 = [1]
cur = 0
for i in range(1, len(x_z4)):
    if x_z4[i] == x_z4[i-1] :
        ans2_z4[cur] += 1
    else:
        cur += 1
        ans2_z4.append(1)

print(ans1_z4)
print(ans2_z4)
```

```python
x_z4 = np.array([2, 2, 2, 3, 3, 3, 5])

start = np.append(np.array([True]), x_z4[1:] != x_z4[:-1])
ind = np.where(start)[0]
val = x_z4[ind]
coun = np.diff(np.append(ind, len(x_z4)))

print((val, coun))
```
